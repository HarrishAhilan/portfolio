### Harrish's code

```python
public class Student {
    String name;
    int grade;
    int pets;
    int siblings;
    String favoriteSport;
    public Student(String name, int grade, int pets, int siblings, String favoriteSport) {
        this.name = name;
        this.grade = grade;
        this.pets = pets;
        this.siblings = siblings;
        this.favoriteSport = favoriteSport;
    }
    public void printInfo() {
        System.out.println(name + " is in grade " + grade + " has " + pets + " pets and " + siblings + " siblings and likes " + favoriteSport);
    }
}
Student student1 = new Student("Harrish", 11, 1, 2, "Basketball");
Student student2 = new Student("Arnav", 10, 0, 1, "Soccer");
Student student3 = new Student("Dhyan", 12, 2, 0, "Tennis");
student1.printInfo();
student2.printInfo();
student3.printInfo();
```
