### Harrish's code

```java
public class Fighter {
    String name;
    int power;
    int health;
    int armor;
    static double fightDuration = 5.5;
    public Fighter(String name, int power, int health, int armor) {
        this.name = name;
        this.power = power;
        this.health = health;
        this.armor = armor;
    }
    void attack(Fighter other) {
        int damage = power - other.armor;
        if (damage < 0) {
            damage = 0;
        }
        other.health -= damage;
        System.out.println(name + " attacks " + other.name + " for " + damage + " damage");
    }
    void printStatus() {
        System.out.println(name + " has " + health + " health and " + power + " power");
    }
    void heal() {
        health += 10;
        System.out.println(name + " heals and now has " + health + " health");
    }
    static int strongerFighter(Fighter a, Fighter b) {
        if (a.power > b.power) {
            return 1;
        } else if (b.power > a.power) {
            return 2;
        } else {
            return 0;
        }
    }
    static void beginBattle(Fighter a, Fighter b) {
        System.out.println("The battle between " + a.name + " and " + b.name + " has begun");
        System.out.println("Fight duration is " + fightDuration + " minutes");
    }
    static void announceWinner(Fighter a, Fighter b) {
        int result = strongerFighter(a, b);
        if (result == 1) {
            System.out.println(a.name + " is the stronger fighter");
        } else if (result == 2) {
            System.out.println(b.name + " is the stronger fighter");
        } else {
            System.out.println("Both fighters are equally strong");
        }
    }
    public static void main(String[] args) {
        Fighter dragon = new Fighter("Dragon", 40, 100, 5);
        Fighter titan = new Fighter("Titan", 35, 120, 8);
        beginBattle(dragon, titan);
        dragon.attack(titan);
        titan.attack(dragon);
        dragon.printStatus();
        titan.printStatus();
        dragon.heal();
        announceWinner(dragon, titan);
    }
}
```
