<h3> Finding a substring within a string </h3>

We can use the "window" method:

A "window" is created the same length as the substring. We then iterate through the main string in sections and compare to the substring

For example:

<h3> I T E R A T E </h3>with substring "ERA"

<br>
<br>
<br>

````python
public class StringFinder {
    public static void main(String[] args) {
        String word = "iterate";
        String sub = "xxx";
        boolean found = false; // will be set to true once substring is found

        for (int i = 0; i < word.length(); i++) { //iterating forwards: starting at first index (0) and going to the length of the word.. let's try word.length
            String portion = word.substring(i, i + sub.length());
            if (sub.length() > portion.length()) break;
            if (portion.equals(sub)) { // make sure you use .equals!!
                found = true;
                break;
            }
        }

        if (found)
            System.out.println("substring is found within string!");
        else
            System.out.println("substring is NOT within string");
    }

}


StringFinder.main(null);
````

### Harrish's answer

The original code crashed with a StringIndexOutOfBoundsException. The loop let i go all the way up to word.length() minus 1, and each step called word.substring(i, i + sub.length()). Once i got close to the end of the word, i + sub.length() went past the actual length of the string, and substring does not allow that. The check comparing sub.length() to portion.length() ran after substring was already called, so it never got the chance to stop the crash. The fix is to change the loop so i only goes up to word.length() - sub.length(), which guarantees the window never reaches past the end of the string. This also correctly handles RATE, since RATE sits right at the very end of ITERATE and the old loop would have crashed trying to check that final window.

### Harrish's code

````python
public class VowelCounter {
    public static void main(String[] args) {
        String word = "iterate";
        int count = 0;
        for (int i = 0; i < word.length(); i++) {
            char c = word.charAt(i);
            if (c == 'a' || c == 'e' || c == 'i' || c == 'o' || c == 'u') {
                count++;
            }
        }
        System.out.println("Number of vowels: " + count);
    }
}
VowelCounter.main(null);
````
